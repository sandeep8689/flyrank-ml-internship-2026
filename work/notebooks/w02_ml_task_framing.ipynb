## My Lane as an ML Task

I chose **Classification** as the machine learning task.

The objective is to predict whether a content item should be refreshed based on its search and engagement metrics. The model classifies each content item into one of two categories: **Needs Refresh** or **Does Not Need Refresh**.

Classification is appropriate because the output is a fixed category rather than a continuous value.

## Target or Proxy

The target is whether a content item should be refreshed.

Since the dataset does not contain a direct "Needs Refresh" label, a proxy target can be created using business rules. For example, content with low CTR, low engagement rate, or poor average position can be labeled as "Needs Refresh," while the remaining content is labeled as "Does Not Need Refresh."

## Why ML Beats a Fixed Rule

Search performance depends on many interacting factors such as CTR, search volume, competition, average position, engagement rate, and content characteristics.

A fixed rule cannot capture these complex relationships. Machine learning learns patterns from historical data and can make more accurate and adaptable decisions than manually written rules.
